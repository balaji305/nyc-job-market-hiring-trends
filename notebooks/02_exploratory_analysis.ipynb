# 02 — Exploratory Hiring Analysis
Posting counts describe advertised roles, while `number_of_positions` describes vacancies. Salary metrics use valid annualized midpoints.

```python
from pathlib import Path
import pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data/processed/jobs_nyc_postings_clean.csv', parse_dates=['posting_date'])
df.shape
```

```python
kpis = {'postings': len(df), 'advertised_positions': df.number_of_positions.sum(), 'agencies': df.agency_clean.nunique(), 'categories': df.job_category_clean.nunique(), 'average_salary': df.salary_midpoint.mean(), 'median_salary': df.salary_midpoint.median()}
kpis
```

```python
df.agency_clean.value_counts().head(10).plot.barh(title='Top Hiring Agencies')
```

```python
df.posting_month.value_counts().sort_index().plot.line(marker='o', title='Postings by Month')
```

```python
df.job_category_clean.value_counts().head(10).to_frame('postings')
```

```python
pay = df.groupby('job_category_clean').salary_midpoint.agg(['count','median','mean'])
pay.query('count >= 10').sort_values('median', ascending=False).head(10)
```

```python
df.experience_level.value_counts().to_frame('postings'), df.posting_type.value_counts().to_frame('postings')
```

## Interpretation guardrails
The dataset is a rolling current-postings inventory, so monthly counts reflect postings still present in the snapshot and should not be interpreted as a complete historical flow. Correlations do not imply hiring outcomes, and no candidate funnel fields are available.
